# Linear Regression From Scratch

Fit a line by minimizing squared errors.

## What to look for

The points follow an upward trend; the line shows the least-squares fit.

This notebook uses the same small example as `main.py`. The chart preview below is embedded so it appears as soon as you open the notebook. Run the code cell to recalculate the printed results and interactive Matplotlib figure (requires NumPy and Matplotlib).


```python
"""Least-squares line fit from basic sums."""
import numpy as np
import matplotlib.pyplot as plt
# These six points are a tiny made-up dataset for practicing a line fit.
x=np.array([1,2,3,4,5,6.],float); y=np.array([1.2,1.9,3.2,3.8,5.1,5.8])
# Least squares chooses the slope that makes squared vertical errors small.
xm,ym=x.mean(),y.mean(); slope=np.sum((x-xm)*(y-ym))/np.sum((x-xm)**2); intercept=ym-slope*xm
pred=slope*x+intercept
print(f"Line: y = {slope:.2f}x + {intercept:.2f}"); print(f"Mean squared error: {np.mean((y-pred)**2):.3f}")
plt.scatter(x,y,label="observations"); plt.plot(x,pred,color="tomato",label="fitted line")
plt.xlabel("x"); plt.ylabel("y"); plt.title("Linear regression from sums"); plt.legend(); plt.grid(alpha=.25)
plt.tight_layout(); plt.show()
```
