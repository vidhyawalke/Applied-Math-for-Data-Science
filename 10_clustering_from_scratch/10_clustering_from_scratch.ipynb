# Clustering From Scratch

Group points by distance using k-means.

## What to look for

Colors show the three groups; crosses show their centers.

This notebook uses the same small example as `main.py`. The chart preview below is embedded so it appears as soon as you open the notebook. Run the code cell to recalculate the printed results and interactive Matplotlib figure (requires NumPy and Matplotlib).


```python
"""Basic k-means clustering with three visible groups."""
import numpy as np
import matplotlib.pyplot as plt
# Each generated cloud forms one visible group in the example data.
rng=np.random.default_rng(10); points=np.vstack([rng.normal([0,0],.7,(30,2)),rng.normal([4,1],.7,(30,2)),rng.normal([2,4],.7,(30,2))])
centers=points[[0,30,60]].copy()
# K-means repeats assignment and center updates until the groups stop moving.
for step in range(30):
    # Assign each point to its nearest center using Euclidean distance.
    distances=np.linalg.norm(points[:,None,:]-centers[None,:,:],axis=2); labels=distances.argmin(1)
    # Replace each center with the mean of its assigned points.
    new=np.array([points[labels==k].mean(0) for k in range(3)])
    if np.allclose(new,centers): break
    centers=new
print(f"Finished after {step+1} updates; centers:\n",np.round(centers,2))
plt.scatter(points[:,0],points[:,1],c=labels,cmap="viridis",s=35); plt.scatter(centers[:,0],centers[:,1],marker="X",s=180,color="tomato",label="centers")
plt.xlabel("x"); plt.ylabel("y"); plt.title("K-means groups points by distance"); plt.legend(); plt.grid(alpha=.2)
plt.tight_layout(); plt.show()
```
